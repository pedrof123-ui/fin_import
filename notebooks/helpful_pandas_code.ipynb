```python
###### IMPORT MULTIPLE FILES FROM A FOLDER ######
import pandas as pd
from pathlib import Path

def load_all_files(folder="incoming"):
    dataframes = []

    for file in Path(folder).glob("*"):
        try:
            if file.suffix == ".csv":
                df = pd.read_csv(file, encoding="utf-8", engine="python")
            elif file.suffix in [".xls", ".xlsx"]:
                df = pd.read_excel(file)
            else:
                continue

            df["source_file"] = file.name
            dataframes.append(df)

        except Exception as e:
            print("Skipped:", file.name, e)

    return pd.concat(dataframes, ignore_index=True)
```

```python
### TO SEND EMAIL REPORTS ######
import smtplib
from email.mime.text import MIMEText

def send_report(report, recipient):
    msg = MIMEText(report)
    msg["Subject"] = "Automated Data Report"
    msg["From"] = "reports@automation.com"
    msg["To"] = recipient

    with smtplib.SMTP("smtp.gmail.com", 587) as server:
        server.starttls()
        server.login("EMAIL", "PASSWORD")
        server.send_message(msg)
```
